**Exemplo 1.2**  

Gera uma onda senoidal discreta de 2 Hz usando MATLAB.
Assume um tempo de amostragem de 0.01 seg. e use pontos suficientes para tornar
a onda senoidal 1 seg. de comprimento; i.e., o tempo total, TT deve ser 1 seg.

```python
import numpy as np
import matplotlib.pyplot as plt

Ts = .01;                   # Define o tempo de amostragem    
TT = 1;                     # Define o tempo total
f = 2;                      # Define a frequência
t = np.arange(0, TT, Ts);   # Gera o vetor de tempo
x = np.sin(2*np.pi*f*t);    # Gera o sinal senoidal desejado

plt.figure(figsize=(8, 5))
plt.plot(t, x, '.k')        # Plota o sinal com pontos pretos
plt.xlabel('Time (sec)')    # Rótulo do eixo x
plt.ylabel('x(t)')          # Rótulo do eixo y
plt.show()                  # Exibe o gráfico
```
